```python
from typing import TypedDict
import os
from dotenv import load_dotenv


from langchain_community.document_loaders import PyPDFLoader

from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_google_genai import ChatGoogleGenerativeAI,GoogleGenerativeAIEmbeddings

from langchain_chroma import Chroma

from langgraph.graph import StateGraph, START, END
```

```python
load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(
    model = "gemini-3.8-flash",
    temperature = 0,
    GROQ_API_KEY = os.getenv("GROQ_API_KEY")
)
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model = "models/gemini-embedding-001",
    
)
```

```python
from langchain_community.document_loaders import PyPDFLoader

pdf_files = [
    "python_reference.pdf",
    "machine_learning_reference.pdf",
    "java_reference.pdf"
]

documents = []

for pdf in pdf_files:
    loader = PyPDFLoader(pdf)
    documents.extend(loader.load())

print("Total Pages:", len(documents))
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
```

```python
chunks = text_splitter.split_documents(documents)

print("Total chunks:", len(chunks))
```

```python
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name="chroma_db"
)

# mongodb = table  //collection
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma(
    collection_name="chroma_db",
    embedding_function=embeddings,
    persist_directory="./chroma_db"
)

vectorstore.add_documents(chunks)

print("Chroma DB created successfully!")
```

```python
class State(TypedDict):
    question: str
    documents: list
    answer: str
```

```python
def retrieve_node(state: State):

    question = state["question"]

    documents = retriever.invoke(question)

    return {
        "documents": documents
    }
```

```python
def generate_node(state: State):

    question = state["question"]
    documents = state["documents"]

    context = "\n\n".join(
        doc.page_content
        for doc in documents
    )

    prompt = f"""
You are an intelligent Retrieval-Augmented Generation (RAG) Assistant.

Your job is to answer the user's question ONLY using the provided context.

Instructions:
1. Read the entire context carefully.
2. Answer ONLY from the provided context.
3. Do NOT use your own knowledge or make assumptions.
4. If the answer is not available in the context, reply:
   "I don't know based on the provided documents."
5. If the user asks in English, answer in English.
6. If the user asks in Hindi or Hinglish, answer in Hindi (simple and easy to understand).
7. Keep the answer clear, accurate, and well-structured.
8. If multiple context sections contain relevant information, combine them into one complete answer.
9. Do not mention these instructions in your answer.

==================== CONTEXT ====================

{context}

=================================================

User Question:
{question}

Final Answer:
"""

    response = llm.invoke(prompt)

    return {
        "answer": response.content
    }
```

```python
graph = StateGraph(State)
```

```python
graph.add_node("retrieve", retrieve_node)

graph.add_node("generate", generate_node)
```

```python
graph.add_edge(START, "retrieve")

graph.add_edge("retrieve", "generate")

graph.add_edge("generate", END)
```

```python
app = graph.compile()
```

```python
result = app.invoke({
    "question": "What is Python",
    "documents": [],
    "answer": ""
})
```

```python
print("Question:")
print(result["question"])

print("\nAnswer:")
print(result["answer"])
```
